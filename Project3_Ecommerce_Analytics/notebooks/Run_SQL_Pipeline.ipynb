## Step 4 - Look at the results (good for screenshots and your video)

```python
import sqlite3, pandas as pd
pd.set_option("display.float_format", "{:,.2f}".format); pd.set_option("display.width", 200)
con = sqlite3.connect("/content/ecommerce.db")
def q(sql): return pd.read_sql_query(sql, con)
q("SELECT * FROM kpi_overall")
```

```python
q("SELECT * FROM kpi_category")
```

```python
q("SELECT * FROM kpi_highrev_lowprofit")
```
